```python
import pandas as pd

file_path = '/content/drive/MyDrive/gym_members_exercise_tracking.txt'

# Charger les données
df = pd.read_csv(file_path, sep=None, engine='python')

# Afficher la liste des colonnes pour faire notre choix
print("Colonnes disponibles dans le dataset :")
print(df.columns.tolist())

# Supposons qu'on souhaite prédire les Calories Brûlées ('Calories_Burned' ou similaire dans le dataset)
# Nous allons identifier dynamiquement la cible potentielle.
target_candidates = [col for col in df.columns if 'calorie' in col.lower() or 'target' in col.lower()]

if target_candidates:
    target_col = target_candidates[0]
else:
    # Par défaut, on prend la dernière colonne si aucune cible évidente n'est détectée
    target_col = df.columns[-1]

# Définir X (caractéristiques) et Y (cible)
Y = df[target_col]
X = df.drop(columns=[target_col])

print(f"\nVariable cible identifiée (Y) : '{target_col}'")
print(f"Variables explicatives identifiées (X) : {X.columns.tolist()}")

# Affichage des dimensions
print(f"\nDimensions de X : {X.shape}")
print(f"Dimensions de Y : {Y.shape}")
```

```python
# تحديد المتغير التابع (Y) والمتغيرات المستقلة (X)
Y = df['Calories_Burned']
X = df.drop(columns=['Calories_Burned'])

# عرض شكل البيانات للتحقق
print("أبعاد المتغيرات المستقلة (X):", X.shape)
print("أبعاد المتغير التابع (Y):", Y.shape)
```

```python
# Affichage des dimensions de X et Y
print(f"Dimensions de X (variables explicatives) : {X.shape}")
print(f"Dimensions de Y (variable cible) : {Y.shape}")

# Nombre total d'individus (lignes)
print(f"Nombre total d'individus dans le Dataset : {df.shape[0]}")

# Affichage des cinq premières lignes du Dataset
print("\nLes cinq premières lignes du Dataset :")
display(df.head(5))
```

```python
from sklearn.model_selection import train_test_split

# Division des données en ensemble d'entraînement (75%) et ensemble de test (25%)
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.25, random_state=42)

# Affichage du nombre d'individus dans chaque ensemble
print(f"Nombre d'individus pour l'entraînement (X_train) : {X_train.shape[0]} (soit {X_train.shape[0] / X.shape[0] * 100:.1f}%)")
print(f"Nombre d'individus pour le test (X_test) : {X_test.shape[0]} (soit {X_test.shape[0] / X.shape[0] * 100:.1f}%)")
```

### Explication de la répartition (75% Entraînement / 25% Test)

La division des données en sous-ensembles d'entraînement et de test est une étape fondamentale en apprentissage automatique pour les raisons suivantes :

1. **Évaluation objective (Généralisation) :** L'objectif principal est de s'assurer que le modèle peut faire des prédictions précises sur de nouvelles données qu'il n'a jamais vues auparavant. En masquant 25 % des données lors de l'entraînement, nous créons un scénario de test réaliste.
2. **Détection du Surapprentissage (Overfitting) :** Si nous testions le modèle sur les mêmes données que celles utilisées pour l'entraîner, le modèle pourrait simplement mémoriser les exemples (surapprentissage). Un bon score sur l'ensemble d'entraînement couplé à un mauvais score sur l'ensemble de test révèle ce problème.
3. **Ratio 75/25 :** Ce ratio est un standard classique. Il offre un bon compromis :
   * **75 % pour l'entraînement** donne suffisamment de données et de diversité au modèle pour qu'il apprenne correctement les relations complexes.
   * **25 % pour le test** fournit un échantillon statistiquement représentatif et robuste pour évaluer précisément la performance finale du modèle.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Identifier les colonnes catégorielles et numériques
categorical_cols = X_train.select_dtypes(include=['object', 'category']).columns.tolist()
numerical_cols = X_train.select_dtypes(exclude=['object', 'category']).columns.tolist()

# Créer le préprocesseur pour encoder les variables catégorielles
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(drop='first', handle_unknown='ignore'), categorical_cols)
    ], remainder='passthrough'
)

# Définir le pipeline avec le préprocesseur et le modèle de Régression Linéaire
model_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', LinearRegression())
])

# Entraîner le modèle
model_pipeline.fit(X_train, Y_train)

# Prédire sur l'ensemble de test
Y_pred = model_pipeline.predict(X_test)

# Calcul des métriques d'évaluation
mae = mean_absolute_error(Y_test, Y_pred)
mse = mean_squared_error(Y_test, Y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(Y_test, Y_pred)

# Affichage des métriques
print("--- Performances du Modèle de Régression ---")
print(f"Mean Absolute Error (MAE) : {mae:.2f}")
print(f"Mean Squared Error (MSE) : {mse:.2f}")
print(f"Root Mean Squared Error (RMSE) : {rmse:.2f}")
print(f"Coefficient de détermination (R²) : {r2:.4f}")
```

### Explication des Métriques d'Évaluation de Régression

Voici comment interpréter les résultats obtenus ci-dessus pour comprendre les performances de notre modèle de prédiction des calories brûlées :

1. **Mean Absolute Error (MAE - Erreur absolue moyenne) :**
   * Elle représente la moyenne des écarts absolus entre les prédictions et les valeurs réelles.
   * *Interprétation :* En moyenne, les prédictions du modèle s'écartent de la valeur réelle de **MAE** calories (en valeur absolue).

2. **Mean Squared Error (MSE - Erreur quadratique moyenne) :**
   * Elle calcule la moyenne des carrés des erreurs. Comme les erreurs sont élevées au carré, cette métrique pénalise fortement les grandes erreurs (les prédictions très éloignées de la réalité).

3. **Root Mean Squared Error (RMSE - Racine de l'erreur quadratique moyenne) :**
   * C'est la racine carrée de la MSE. L'avantage du RMSE est qu'il s'exprime dans la même unité que la variable cible (les calories).
   * *Interprétation :* Elle montre l'écart type moyen des résidus de prédiction. Une valeur proche de la MAE indique que les erreurs sont réparties de manière homogène sans trop de valeurs aberrantes (grandes erreurs).

4. **Coefficient de détermination ($R^2$) :**
   * Il mesure la proportion de la variance de la variable cible (les calories brûlées) qui est expliquée par les variables prédictives de notre modèle.
   * *Interprétation :* Un score proche de $1$ (ou $100\%$) indique que le modèle explique extrêmement bien les variations des données. Par exemple, si le $R^2$ est de $0.95$, cela signifie que $95\%$ de la variation des calories brûlées est expliquée par les caractéristiques de l'entraînement et le profil physique des membres.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# 1. Performances avec 100% des données d'entraînement (déjà calculées)
perf_100 = {
    "MAE": mae,
    "MSE": mse,
    "RMSE": rmse,
    "R²": r2
}

# 2. Diviser l'ensemble d'entraînement existant pour en obtenir 50%
# On utilise random_state=42 pour assurer la reproductibilité
X_train_50, _, Y_train_50, _ = train_test_split(
    X_train, Y_train, test_size=0.5, random_state=42
)

# 3. Ré-entraînement du modèle sur 50% des données d'entraînement
model_pipeline.fit(X_train_50, Y_train_50)

# Prédictions sur l'ensemble de test
Y_pred_50 = model_pipeline.predict(X_test)

# Calcul des métriques pour le modèle entraîné à 50%
mae_50 = mean_absolute_error(Y_test, Y_pred_50)
mse_50 = mean_squared_error(Y_test, Y_pred_50)
rmse_50 = np.sqrt(mse_50)
r2_50 = r2_score(Y_test, Y_pred_50)

perf_50 = {
    "MAE": mae_50,
    "MSE": mse_50,
    "RMSE": rmse_50,
    "R²": r2_50
}

# 4. Construction et affichage du tableau de comparaison
df_compare = pd.DataFrame({
    "Métrique": ["MAE (Calories)", "MSE", "RMSE (Calories)", "R²"],
    "Modèle 100% (729 individus)": [perf_100["MAE"], perf_100["MSE"], perf_100["RMSE"], perf_100["R²"]],
    "Modèle 50% (364 individus)": [perf_50["MAE"], perf_50["MSE"], perf_50["RMSE"], perf_50["R²"]]
})

display(df_compare)
```

- **Le maintien du pouvoir explicatif ($R^2$) :** Si le modèle conserve une bonne capacité de généralisation malgré la réduction de taille.
